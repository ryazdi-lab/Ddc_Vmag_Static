```python
import numpy as np

Root = "/home/jupyter-r.yazdi/myproject/dti3d/0pycodes/Shahr/Vmag_prj/V_MAG_dataset/"
files = {
    14:  (Root+"case14/dataset_train.npz",  Root+"case14/dataset_test.npz"),
    30:  (Root+"case30/dataset_train.npz",  Root+"case30/dataset_test.npz"),
    39:  (Root+"case39/dataset_train.npz",  Root+"case39/dataset_test.npz"),
    57:  (Root+"case57/dataset_train.npz",  Root+"case57/dataset_test.npz"),
    118: (Root+"case118/dataset_train.npz", Root+"case118/dataset_test.npz"),
    300: (Root+"case300/dataset_train.npz", Root+"case300/dataset_test.npz"),
}

def load_vmag(path, n):
    a = np.load(path)["Vmag"]
    assert a.ndim == 2, f"unexpected shape {a.shape}"
    if a.shape[0] == n:
        return a.T              # stored (N, S) -> (S, N)
    if a.shape[1] == n:
        return a                # already (S, N)
    raise ValueError(f"{path}: shape {a.shape} has no axis of size {n}")

for n, (tr, te) in files.items():
    v_train = load_vmag(tr, n)  # (S_train, N)
    v_test  = load_vmag(te, n)  # (S_test,  N)
    mu = v_train.mean(axis=0)   # per-bus training mean, shape (N,)
    rmse = np.sqrt(np.mean((v_test - mu) ** 2))
    print(f"{n:>4} buses: train {v_train.shape[0]}, test {v_test.shape[0]}, "
          f"per-bus mean range [{mu.min():.4f}, {mu.max():.4f}], RMSE = {rmse:.6f}")
```
